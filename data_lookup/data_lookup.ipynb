```python
import pandas as pd

df = pd.read_csv("../olist_public_dataset/olist_orders_dataset.csv")
df.head(20)
```

```python
# Fully duplicated rows
print("Duplicate rows:", df.duplicated().sum())

# Duplicates by order_id (should be unique)
print("Duplicate order_id:", df.duplicated(subset="order_id").sum())

# customer_id should also be unique per order in this table
print("Duplicate customer_id:", df.duplicated(subset="customer_id").sum())

# Fully duplicated rows, grouped together (first 30)
print("Fully duplicated rows (grouped):")
display(df[df.duplicated(keep=False)].sort_values(list(df.columns)).head(30))

# Rows sharing an order_id, grouped together (first 30)
print("Rows sharing an order_id (grouped):")
display(df[df.duplicated(subset="order_id", keep=False)].sort_values("order_id").head(30))
```

```python
# Missing values per column
missing = df.isnull().sum().to_frame("missing_count")
missing["missing_pct"] = (missing["missing_count"] / len(df) * 100).round(2)
missing
```

```python
# Number of unique values per column
df.nunique().to_frame("unique_count")
```
